# Question 1 Employee Data Bank.
Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python
class Employee:
    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, year_of_experience):

        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print("\n----- Employee Details -----")
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Year of Experience:", self.year_of_experience)

    def update_year_of_experience(self, experience):
        self.year_of_experience = experience
        print("Experience updated successfully!")

    def increase_salary(self, increment):
        self.salary += increment
        print("Salary increased successfully!")

employee101 = Employee(
    "Ali Khan", 101, "Karachi", 50000,
    "Graphic Designer", 2023, "Bachelor's", 2
)

print("Name:", employee101.name)
print("Number:", employee101.number)
print("Address:", employee101.address)
print("Salary:", employee101.salary)

employee101.describe_employee()
employee101.increase_salary(5000)
employee101.update_year_of_experience(3)

employee101.describe_employee()

employee102 = Employee("Sara", 102, "Lahore", 60000, "UI Designer", 2022, "BSCS", 3)
employee103 = Employee("Usman", 103, "Islamabad", 70000, "Developer", 2021, "BSSE", 4)
employee104 = Employee("Ayesha", 104, "Karachi", 55000, "Content Writer", 2024, "MA English", 1)
employee105 = Employee("Hamza", 105, "Multan", 65000, "Video Editor", 2022, "Bachelor's", 3)
employee106 = Employee("Zainab", 106, "Hyderabad", 58000, "SMM", 2023, "BBA", 2)
employee107 = Employee("Bilal", 107, "Peshawar", 80000, "Web Developer", 2020, "BSCS", 5)
employee108 = Employee("Hira", 108, "Quetta", 62000, "Illustrator", 2022, "Bachelor's", 3)
employee109 = Employee("Danish", 109, "Faisalabad", 75000, "Data Analyst", 2021, "BS Statistics", 4)
employee110 = Employee("Maham", 110, "Rawalpindi", 52000, "Graphic Designer", 2024, "Bachelor's", 1)

employees = [
    employee101, employee102, employee103, employee104,
    employee105, employee106, employee107, employee108,
    employee109, employee110
]

for employee in employees:
    employee.describe_employee()

employee102.increase_salary(4000)
employee102.update_year_of_experience(4)

employee105.increase_salary(6000)
employee105.update_year_of_experience(4)

employee110.increase_salary(3000)
employee110.update_year_of_experience(2)

print("\n===== Updated Employee Details =====")
employee102.describe_employee()
employee105.describe_employee()
employee110.describe_employee()
```
